```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# ===========================
# Load CSV
# ===========================
csv_path = "/home/dnarsipu/Documents/Personal/EE5522/Final Project/Dhanush_Semi_UIR/training_metrics.csv"   # <-- replace with your path
df = pd.read_csv(csv_path)

# Create folder for saving plots
save_dir = "Final Project/Dhanush_Semi_UIR/plots/training_plots"
os.makedirs(save_dir, exist_ok=True)

epochs = df["epoch"]

# ===========================
# 1. Total Loss vs Epoch
# ===========================
plt.figure(figsize=(8,5))
plt.plot(epochs, df["main_loss"], label="Total Loss", linewidth=2)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Total Loss vs Epoch")
plt.grid(True)
plt.tight_layout()
plt.savefig(f"{save_dir}/total_loss.png", dpi=300)
plt.close()

# ===========================
# 2. Supervised Loss vs Epoch
# ===========================
plt.figure(figsize=(8,5))
plt.plot(epochs, df["Ls"], label="Supervised Loss (Ls)", color="tab:blue", linewidth=2)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Supervised Loss vs Epoch")
plt.grid(True)
plt.tight_layout()
plt.savefig(f"{save_dir}/supervised_loss.png", dpi=300)
plt.close()

# ===========================
# 3. Unsupervised Loss vs Epoch
# ===========================
plt.figure(figsize=(8,5))
plt.plot(epochs, df["Lu"], label="Unsupervised Loss (Lu)", color="tab:orange", linewidth=2)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Unsupervised Loss vs Epoch")
plt.grid(True)
plt.tight_layout()
plt.savefig(f"{save_dir}/unsupervised_loss.png", dpi=300)
plt.close()

# ===========================
# 4. Training PSNR vs Epoch
# ===========================
plt.figure(figsize=(8,5))
plt.plot(epochs, df["train_psnr"], label="Train PSNR", color="tab:green", linewidth=2)
plt.xlabel("Epoch")
plt.ylabel("PSNR (dB)")
plt.title("Training PSNR vs Epoch")
plt.grid(True)
plt.tight_layout()
plt.savefig(f"{save_dir}/train_psnr.png", dpi=300)
plt.close()

# ===========================
# 5. Validation PSNR vs Epoch
# ===========================
plt.figure(figsize=(8,5))
plt.plot(epochs, df["val_psnr_eval"], label="Validation PSNR", color="tab:red", linewidth=2)
plt.xlabel("Epoch")
plt.ylabel("PSNR (dB)")
plt.title("Validation PSNR vs Epoch")
plt.grid(True)
plt.tight_layout()
plt.savefig(f"{save_dir}/val_psnr.png", dpi=300)
plt.close()

# ===========================
# 6. Validation SSIM vs Epoch
# ===========================
plt.figure(figsize=(8,5))
plt.plot(epochs, df["val_ssim_eval"], label="Validation SSIM", color="tab:purple", linewidth=2)
plt.xlabel("Epoch")
plt.ylabel("SSIM")
plt.title("Validation SSIM vs Epoch")
plt.grid(True)
plt.tight_layout()
plt.savefig(f"{save_dir}/val_ssim.png", dpi=300)
plt.close()

# ===========================
# 7. Learning Rate vs Epoch
# ===========================
plt.figure(figsize=(8,5))
plt.plot(epochs, df["lr"], label="Learning Rate", color="black", linewidth=2)
plt.xlabel("Epoch")
plt.ylabel("Learning Rate")
plt.title("Learning Rate Schedule")
plt.grid(True)
plt.tight_layout()
plt.savefig(f"{save_dir}/learning_rate.png", dpi=300)
plt.close()

# ===========================
# 8. Combined Grid Figure (optional)
# ===========================
fig, axs = plt.subplots(3, 3, figsize=(18, 12))
axs = axs.flatten()

# List of curves to plot
plots = [
    ("Total Loss", df["main_loss"]),
    ("Supervised Loss", df["Ls"]),
    ("Unsupervised Loss", df["Lu"]),
    ("Train PSNR", df["train_psnr"]),
    ("Val PSNR", df["val_psnr_eval"]),
    ("Val SSIM", df["val_ssim_eval"]),
    ("Learning Rate", df["lr"])
]

for i, (title, y) in enumerate(plots):
    axs[i].plot(epochs, y, linewidth=2)
    axs[i].set_title(title)
    axs[i].set_xlabel("Epoch")
    axs[i].set_ylabel(title)
    axs[i].grid(True)

# Hide last empty plot
axs[-1].axis("off")

plt.tight_layout()
plt.savefig(f"{save_dir}/training_curves_grid.png", dpi=300)
plt.close()

print("All plots saved in:", save_dir)
```
